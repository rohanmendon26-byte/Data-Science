```python
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Math': [85, 78, 92],
    'Science': [90, 82, 89],
    'English': [88, 85, 94]
}

df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

```python
df.melt(id_vars=["Name"],value_vars=["Math","Science","English"],var_name="Subject",value_name="Score")
```

```python
import pandas as pd

data1 = {
    "Name": [
        "Alice", "Alice", "Alice",
        "Bob", "Bob", "Bob",
        "Charlie", "Charlie", "Charlie"
    ],
    "Subject": [
        "Math", "Science", "English",
        "Math", "Science", "English",
        "Math", "Science", "English"
    ],
    "Score": [
        85, 90, 88,
        78, 82, 85,
        92, 89, 94
    ]
}

df = pd.DataFrame(data1)

print(df)
```

```python
df.pivot(index="Name", columns="Subject", values="Score")
```

```python
import pandas as pd

data = {
    "Name": [
        "Alice",
        "Alice",
        "Alice",
        "Bob",
        "Bob"
    ],
    "Subject": [
        "Math",
        "Math",
        "Science",
        "Math",
        "Math"
    ],
    "Score": [
        85,
        80,
        90,
        78,
        82
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.pivot_table(index="Name", columns="Subject", values="Score", aggfunc="mean")
```
